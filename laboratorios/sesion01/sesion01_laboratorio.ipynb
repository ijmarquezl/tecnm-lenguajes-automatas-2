### Fase 1: El Artesano (15 min - Sin IA)
**Reto:** Completa la función `invertir_cadena()` usando únicamente aritmética de punteros `inicio` y `fin` *in-place* (sin arreglos auxiliares ni `malloc`).

```python
%%writefile reto_artesano_c.c
#include <stdio.h>
#include <string.h>

void invertir_cadena(char *str) {
    if (str == NULL || *str == '\0') return;
    
    char *inicio = str;
    char *fin = str + strlen(str) - 1;
    char temp;

    // TODO (El Artesano): Implementa el bucle de intercambio manual con punteros
    while (inicio < fin) {
        temp = *inicio;
        *inicio = *fin;
        *fin = temp;
        inicio++;
        fin--;
    }
}

int main() {
    char buffer[] = "COMPILADOR_SCD1016";
    printf("[C - ORIGINAL]  %s\n", buffer);
    invertir_cadena(buffer);
    printf("[C - INVERTIDA] %s\n", buffer);
    return 0;
}
```

### Fase 1: El Artesano (15 min - Sin IA)
**Reto:** Implementa la función `invertir_bytes()` manipulando un slice mutable de bytes (`&mut [u8]`) sin reasignar memoria en el *Heap*.

```python
%%writefile reto_artesano_rs.rs
fn invertir_bytes(buffer: &mut [u8]) {
    if buffer.is_empty() { return; }
    let mut i = 0;
    let mut j = buffer.len() - 1;
    
    // TODO (El Artesano): Intercambio manual de índices
    while i < j {
        buffer.swap(i, j);
        i += 1;
        j -= 1;
    }
}

fn main() {
    let mut buffer = *b"COMPILADOR_SCD1016";
    println!("[RUST - ORIGINAL]  {:?}", std::str::from_utf8(&buffer).unwrap());
    invertir_bytes(&mut buffer);
    println!("[RUST - INVERTIDA] {:?}", std::str::from_utf8(&buffer).unwrap());
}
```

### Fase 2: El Científico — Track Rust
1. Define la matriz formal de requerimientos considerando las reglas del *Borrow Checker* y mutabilidad exclusiva.
2. Diseña tu prompt estructurado solicitando código idiomático en Rust.
3. Pega el código devuelto por la IA en la siguiente celda:

```python
%%writefile codigo_ia_rs.rs
// PEGA AQUÍ EL CÓDIGO GENERADO POR LA IA PARA EL TRACK RUST
pub fn invertir_bytes_ia(buffer: &mut [u8]) {
    if buffer.is_empty() { return; }
    let mut inicio = 0;
    let mut fin = buffer.len() - 1;
    while inicio < fin {
        buffer.swap(inicio, fin);
        inicio += 1;
        fin -= 1;
    }
}
```

### Fase 3: El Auditor — Test Harness en Rust
Ejecuta la suite de pruebas unitarias nativas de Rust (`#[test]`):

```python
%%writefile test_harness_rs.rs
mod codigo_ia_rs;
use codigo_ia_rs::invertir_bytes_ia;

fn main() {
    println!("=== AUDITORIA CRITICA: TRACK RUST ===");

    // Test 1: Slice vacío
    print!("[TEST 1] Buffer vacio... ");
    let mut vacio: [u8; 0] = [];
    invertir_bytes_ia(&mut vacio);
    assert_eq!(vacio, []);
    println!("PASO ✅");

    // Test 2: Un solo elemento
    print!("[TEST 2] Un solo elemento [b'X']... ");
    let mut un_elem = *b"X";
    invertir_bytes_ia(&mut un_elem);
    assert_eq!(&un_elem, b"X");
    println!("PASO ✅");

    // Test 3: Longitud par
    print!("[TEST 3] Longitud par b\"HOLA\"... ");
    let mut par = *b"HOLA";
    invertir_bytes_ia(&mut par);
    assert_eq!(&par, b"ALOH");
    println!("PASO ✅");

    // Test 4: Longitud impar
    print!("[TEST 4] Longitud impar b\"RADAR\"... ");
    let mut impar = *b"RADAR";
    invertir_bytes_ia(&mut impar);
    assert_eq!(&impar, b"RADAR");
    println!("PASO ✅");

    println!("\nRESULTADO RUST: Todas las pruebas estructurales pasaron con exito.\n");
}
```
