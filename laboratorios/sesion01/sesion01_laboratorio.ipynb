### Fase 1: El Artesano (15 min - Sin IA)
**Reto Diagnóstico: Recorrido e Inversión con Apuntadores**

En compiladores, un programa fuente es una secuencia de caracteres en memoria (`char*`). Para procesarlo, manipulamos direcciones y desplazamientos constantes.

**Objetivo:**
Completa la función `invertir_cadena(char *inicio)` en C (o su equivalente en Rust) para invertir una cadena de texto *in-place* (sin usar memoria auxiliar ni arreglos adicionales), manipulando **exclusivamente dos apuntadores**: uno al inicio y otro al final de la cadena.

```python
%%writefile reto_artesano.c
#include <stdio.h>
#include <string.h>

void invertir_cadena(char *str) {
    if (str == NULL || *str == '\0') return;
    
    char *inicio = str;
    char *fin = str + strlen(str) - 1;
    char temp;

    // TODO: Escribe aquí el bucle de intercambio usando exclusivamente los apuntadores 'inicio' y 'fin'
    while (inicio < fin) {
        temp = *inicio;
        *inicio = *fin;
        *fin = temp;
        
        inicio++;
        fin--;
    }
}

int main() {
    char buffer[] = "COMPILADOR_SCD1016";
    
    printf("[ANTES] Cadena original: %s\n", buffer);
    printf("[MEM]   Direccion base:   %p\n", (void*)buffer);
    
    invertir_cadena(buffer);
    
    printf("[DESPUES] Cadena invertida: %s\n", buffer);
    return 0;
}
```

```python
# Compilar con advertencias estrictas y ejecutar binario en C
!gcc -Wall -Wextra reto_artesano.c -o reto_artesano && ./reto_artesano
```

### Track Alternativo en Rust (Opcional para alumnos del track Rust)

```python
%%writefile reto_artesano.rs
fn invertir_bytes(buffer: &mut [u8]) {
    if buffer.is_empty() { return; }
    let mut i = 0;
    let mut j = buffer.len() - 1;
    while i < j {
        buffer.swap(i, j);
        i += 1;
        j -= 1;
    }
}

fn main() {
    let mut buffer = *b"COMPILADOR_SCD1016";
    println!("[ANTES] {:?}", std::str::from_utf8(&buffer).unwrap());
    invertir_bytes(&mut buffer);
    println!("[DESPUES] {:?}", std::str::from_utf8(&buffer).unwrap());
}
```

```python
# Compilar y ejecutar binario en Rust
!rustc reto_artesano.rs -o reto_rust && ./reto_rust
```
