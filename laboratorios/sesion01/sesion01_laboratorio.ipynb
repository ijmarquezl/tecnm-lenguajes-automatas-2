### Fase 1: El Artesano (15 min - Sin IA)
**Reto:** Completa la función `invertir_cadena()` usando únicamente aritmética de punteros `inicio` y `fin`.

```python
%%writefile reto_artesano.c
#include <stdio.h>
#include <string.h>

void invertir_cadena(char *str) {
    if (str == NULL || *str == '\0') return;
    
    char *inicio = str;
    char *fin = str + strlen(str) - 1;
    char temp;

    // TODO (El Artesano): Implementa el bucle de intercambio manual
    while (inicio < fin) {
        temp = *inicio;
        *inicio = *fin;
        *fin = temp;
        inicio++;
        fin--;
    }
}

int main() {
    char buffer[] = "COMPILADOR_SCD1016";
    printf("[ORIGINAL]  %s\n", buffer);
    invertir_cadena(buffer);
    printf("[INVERTIDA] %s\n", buffer);
    return 0;
}
```

### Fase 2: El Científico (Diseño de Requerimientos y Prompting)
1. Define la matriz de precondiciones, postcondiciones y casos límite en tu bitácora.

3. Guarda el código propuesto por la IA en el archivo `codigo_ia.c` para someterlo a pruebas.

```python
%%writefile codigo_ia.c
#include <stdio.h>
#include <string.h>

// PEGA AQUÍ EL CÓDIGO SUGERIDO POR LA IA PARA AUDITARLO
void invertir_cadena_ia(char *str) {
    if (!str) return;
    char *inicio = str;
    char *fin = str + strlen(str) - 1;
    while (inicio < fin) {
        char t = *inicio;
        *inicio++ = *fin;
        *fin-- = t;
    }
}
```

### Fase 3: El Auditor (Test Harness Automatizado)
Ejecuta la siguiente suite de pruebas destructivas sobre el código propuesto por la IA para detectar fallos de segmentación o casos límite ignorados.

```python
%%writefile test_harness.c
#include <stdio.h>
#include <string.h>
#include <assert.h>

// Prototipo de la función auditada
void invertir_cadena_ia(char *str);

int main() {
    printf("=== INICIANDO BATERIA DE AUDITORIA CRITICA ===\n");

    // Test 1: Puntero NULL
    printf("[TEST 1] Puntero NULL... ");
    invertir_cadena_ia(NULL);
    printf("PASO ✅\n");

    // Test 2: Cadena vacia
    printf("[TEST 2] Cadena vacia \"\"... ");
    char vacia[] = "";
    invertir_cadena_ia(vacia);
    assert(strcmp(vacia, "") == 0);
    printf("PASO ✅\n");

    // Test 3: Un solo caracter
    printf("[TEST 3] Un solo caracter \"X\"... ");
    char un_car[] = "X";
    invertir_cadena_ia(un_car);
    assert(strcmp(un_car, "X") == 0);
    printf("PASO ✅\n");

    // Test 4: Longitud Par
    printf("[TEST 4] Longitud par \"HOLA\"... ");
    char par[] = "HOLA";
    invertir_cadena_ia(par);
    assert(strcmp(par, "ALOH") == 0);
    printf("PASO ✅\n");

    // Test 5: Longitud Impar
    printf("[TEST 5] Longitud impar \"RADAR\"... ");
    char impar[] = "RADAR";
    invertir_cadena_ia(impar);
    assert(strcmp(impar, "RADAR") == 0);
    printf("PASO ✅\n");

    printf("\nRESULTADO: Todas las pruebas estructurales pasaron con exito.\n");
    return 0;
}
```
